```python
# adding annotatons with text function
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label ="Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.text(10,95,"This value is important!")
```

```python
# adding annotatons with text function and modifyng the text position and colour
# using the ha parameter to change the text position
# and the color parameter to change the text colour
# using the size parameter to change the text size
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label ="Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.text(10,95,"This value is important!",
        ha = 'center',
        color = 'red',
        size = 20)
```

```python
# load libraries
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

# sample data 
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0,75, 50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
# introducing subplots
fig, (ax1, ax2) = plt.subplots(ncols=2,
                               nrows = 1,
                               figsize = (7,3))

ax1.scatter(x1,y1)
ax2.bar(x2,y2)
```

```python
# changing the colors of the subplots
fig, (ax1, ax2) = plt.subplots(ncols=2,
                               nrows = 1,
                               figsize = (7,3))

ax1.scatter(x1,y1, color='pink')
ax2.bar(x2,y2,color='indigo')

## activity add at least one modification to your subplots 
```

```python
# Defining the subplot using subplot mosaic
# using the subplot_mosaic function to define the subplots
# and the layout of the subplots
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize = (7,4))

someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
```

```python
# adding labels to the subplots
# using the set_xlabel function to add labels to the subplots
# and changing the font size of the labels
# added the layout parameter to the subplot_mosaic function to change the layout of the subplots
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize = (7,4),
                                    layout = 'constrained')

someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)

someaxes['ax1'].set_xlabel('A Big Label', fontsize = 18)
someaxes['ax2'].set_xlabel('Another Label', fontsize = 18)
someaxes['ax3'].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize = 18)
```

```python
# making multiple visualisations in one axes object 
x = np.array(['Luffy', 'Zoro', 'Nami', 'Usopp', 'Sanji'])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

fig, ax = plt.subplots(figsize = (7,3))

ax.bar(x,y1, color='indigo')
ax.plot(x,y2, color='red')
```

```python
# adding error bars to the plot
# using the errorbar function to add error bars to the plot
# and changing the color of the error bars
# using the yerr parameter to set the error bars


y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize = (7,3))
ax.plot(x,y2, color='red')

ax.errorbar(x, y2,  fmt="none")
```

```python
# customising the error bars
# using the errorbar function to add error bars to the plot
# and the ecolor parameter to set the color of the error bars
# using the elinewidth parameter to set the width of the error bars
# using the capsize parameter to set the size of the caps on the error bars
# using the capthick parameter to set the thickness of the caps on the error bars
# using the errorevery parameter to set the number of error bars to be displayed
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize = (7,3))
ax.plot(x,y2, color='red')

ax.errorbar(x, y2, yerr=y2_sd, fmt="none",
            ecolor='indigo',
            elinewidth = 4,
            capsize = 6,
            capthick = 4,
            errorevery = 3)
```

### Seaborn Visualization 

```python
# load libraries and packages 

import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# load data for use in seaborn
# using the load_dataset function to load the tips dataset

tips = sns.load_dataset("tips")
print(tips)
```

```python
# making a simple plot using seaborn and changing the style
# using the set_style function to change the style of the plot
#sns.set_style('darkgrid')
sns.set_style('whitegrid')
sns.lineplot(data=tips, x='total_bill', y='tip')
```

```python
# Modifying the aesthetics of the plot
fig = plt.subplots(figsize = (10,3))

tipgraph = sns.lineplot(data=tips, x='total_bill', y='tip',
                        color = 'hotpink',
                        linestyle = '--',
                        linewidth = 3,
                        marker = 'o',
                        markerfacecolor = 'indigo')

tipgraph.set(title = 'Tips vs Total Bill',
             xlabel = 'Total Bill ($)',
             ylabel = 'Tip Amount ($)')
```

```python
# adding multiple variables to the plot
# using the style and hue parameters to add multiple variables to the plot
tipgraph = sns.scatterplot(data=tips, x='total_bill', y='tip',
                           style = 'time', hue = 'day',
                           palette = ['purple', 'hotpink', 'deepskyblue',
                                      'yellowgreen'])

tipgraph.set(title = 'Tips vs Total Bill',
             xlabel = 'Total Bill',
             ylabel = 'Tip Amount')
```

```python
# Pair plotting 
sns.pairplot(data=tips, hue='day')
```

```python
# trying out replot
daysplot = sns.relplot(data = tips, x='total_bill', y='tip',
                       hue='sex', col='day', 
                       kind='scatter', col_wrap=2)
```

Word clouds and venn diagrams

```python
# loading the datset of movie quotes
# using the read_csv function to load the dataset
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv", 
                 on_bad_lines='skip')

df      
```

```python
# Adding modifications to the plot 
text = " ".join(each for each in df.quote)

wordcloud = WordCloud(background_color='white',
                      colormap='inferno').generate(text)

fig, ax = plt.subplots(figsize = (7,3))
ax.imshow(wordcloud, interpolation='bilinear')
ax.axis('off')
```

```python
# making venn diagrams
A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", 'blueberry', 'apple', 'key lime'])
```

```python
diagram = venn2_unweighted([A, B],
                         set_labels = ('Fruits', 'Pies'),
                         set_colors=('blue', 'red'),
                         alpha = 0.5)


diagram.get_label_by_id('10').set_text('\n'.join(A-B))
diagram.get_label_by_id('11').set_text('\n'.join(A&B))
diagram.get_label_by_id('01').set_text('\n'.join(B-A))
```
